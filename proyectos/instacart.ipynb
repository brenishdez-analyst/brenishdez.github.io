### `order_products` data frame

```python
print(order_products.isnull().sum())# Encuentra los valores ausentes
```

```python
print("Mínimo add_to_cart_order:", order_products['add_to_cart_order'].min())
print("Máximo add_to_cart_order:", order_products['add_to_cart_order'].max())# ¿Cuáles son los valores mínimos y máximos en esta columna?
```

Describe brevemente cuáles son tus hallazgos.

En order_products, los ausentes aparecen en add_to_cart_order (orden de adición al carrito). El mínimo observado es 1 y el máximo corresponde al tamaño del carrito más grande registrado (64).

```python
ids_faltantes = order_products[order_products['add_to_cart_order'].isnull()]['order_id']

print("Número de pedidos con add_to_cart_order ausente:", ids_faltantes.nunique())
print("Algunos IDs con valor ausente:")
print(ids_faltantes.unique()[:10])

ids_faltantes_lista = ids_faltantes.unique().tolist()# Guarda todas las IDs de pedidos que tengan un valor ausente en 'add_to_cart_order'
```

```python
pedidos_con_na = order_products[order_products['order_id'].isin(ids_faltantes_lista)]

conteo_productos = pedidos_con_na.groupby('order_id').size() # Agrupa todos los pedidos con datos ausentes por su ID de pedido.

print("Mínimo de productos en pedidos con add_to_cart_order ausente:", int(conteo_productos.min()))
print("Máximo de productos en pedidos con add_to_cart_order ausente:", int(conteo_productos.max()))

todos_mas_64 = (conteo_productos > 64).all()
print("¿Todos los pedidos con valores ausentes tienen más de 64 productos?:", bool(todos_mas_64))
```

Describe brevemente cuáles son tus hallazgos.

Al agrupar los pedidos con valores ausentes en add_to_cart_order, se observó que todos estos pedidos tienen un tamaño inusualmente grande en comparación con el promedio. El conteo mínimo de productos en dichos pedidos supera el umbral de 64 productos, lo que confirma que los valores ausentes están relacionados con pedidos muy extensos.

Esto sugiere que los NaN en add_to_cart_order podrían deberse a un error en la asignación del orden de los artículos cuando el carrito sobrepasa cierta cantidad de productos. Por esta razón, en el preprocesamiento se imputará el valor ausente con un marcador razonable (por ejemplo, el máximo posible dentro de ese pedido) o con 1 como alternativa simple, asegurando que la fila no se pierda.

```python

order_products['add_to_cart_order'] = order_products['add_to_cart_order'].fillna(999)

order_products['add_to_cart_order'] = order_products['add_to_cart_order'].astype(int)

print("Valores ausentes:", order_products['add_to_cart_order'].isnull().sum())
print("Tipo de dato:", order_products['add_to_cart_order'].dtype)
order_products.head()# Remplaza los valores ausentes en la columna 'add_to_cart? con 999 y convierte la columna al tipo entero.

```

Describe brevemente tus hallazgos y lo que hiciste con ellos.

En el DataFrame order_products, los valores ausentes en la columna add_to_cart_order se reemplazaron por el valor 999. Esta imputación funciona como un marcador artificial, ya que dicho número está fuera del rango esperado de la secuencia real de adición al carrito.

Posteriormente, la columna fue convertida al tipo entero, asegurando la consistencia de los datos y facilitando su uso en cálculos posteriores.

```python

miercoles = orders[orders['order_dow'] == 3]['order_hour_of_day'].value_counts().sort_index()
sabado = orders[orders['order_dow'] == 6]['order_hour_of_day'].value_counts().sort_index()

ax = miercoles.plot(kind='bar', figsize=(10,6), color='skyblue', alpha=0.7, label="Miércoles")
sabado.plot(kind='bar', color='orange', alpha=0.7, label="Sábado", ax=ax)

plt.title("Distribución de pedidos por hora del día (Miércoles vs Sábado)")
plt.xlabel("Hora del día (0–23)")
plt.ylabel("Número de pedidos")
plt.legend()
plt.show()
```

El gráfico muestra que las distribuciones de pedidos en miércoles y sábados presentan similitudes, pero también algunas diferencias:

En miércoles, la actividad de pedidos se concentra en horas laborales, con picos importantes entre las 10:00 y 17:00 horas.

En sábado, los pedidos tienden a repartirse más tarde en el día, con mayor presencia en las tardes y noches, lo que refleja hábitos de compra distintos durante el fin de semana.

Ambos días muestran poca actividad durante la madrugada (0–6 horas).

En conclusión, los clientes mantienen un patrón de compras frecuente en ambos días, pero con horarios de mayor concentración laboral entre semana y horarios más flexibles los fines de semana.

### [C3] Para cada producto, ¿cuál es la proporción de las veces que se pide y que se vuelve a pedir?

```python
total_por_producto = order_products.groupby('product_id')['reordered'].count()
```

```python
reordenado_por_producto = order_products.groupby('product_id')['reordered'].sum()
```

```python
tasa_reordenamiento = (reordenado_por_producto / total_por_producto).reset_index()
tasa_reordenamiento.columns = ['product_id', 'tasa_reordenamiento']

tasa_reordenamiento = tasa_reordenamiento.merge(products[['product_id', 'product_name']], on='product_id')

print(tasa_reordenamiento.head(20))
```

La tasa de reordenamiento es un indicador clave para identificar los productos esenciales y de compra repetitiva frente a los productos de compra ocasional.

### [C4] Para cada cliente, ¿qué proporción de sus productos ya los había pedido?

```python
op_users = order_products.merge(orders[['order_id','user_id']], on='order_id')
totales_usuario = op_users.groupby('user_id')['reordered'].count()
reordenados_usuario = op_users.groupby('user_id')['reordered'].sum()
```

```python
tasa_reorden_usuario = (reordenados_usuario / totales_usuario).reset_index()
tasa_reorden_usuario.columns = ['user_id','tasa_reordenamiento']

print(tasa_reorden_usuario.head())
```

Instacart cuenta con perfiles de clientes muy variados: clientes exploradores que diversifican sus compras y clientes leales que repiten casi siempre los mismos productos. Esta métrica es útil para segmentar usuarios y personalizar recomendaciones.

### [C5] ¿Cuáles son los 20 principales artículos que las personas ponen primero en sus carritos?

```python
primeros_productos = order_products[order_products['add_to_cart_order'] == 1]
```

```python
top20_primeros = primeros_productos['product_id'].value_counts().head(20)
```

```python
top20_primeros = products[products['product_id'].isin(top20_primeros.index)][['product_id', 'product_name']] \
    .merge(top20_primeros.rename("veces_primer_articulo"), left_on="product_id", right_index=True) \
    .sort_values("veces_primer_articulo", ascending=False)

print(top20_primeros)
```

El análisis muestra que los 20 artículos más comunes como primer producto en el carrito corresponden principalmente a productos básicos de alta prioridad en la compra, como leche, frutas, pan o café.

Esto refleja que los clientes tienden a comenzar sus pedidos con productos esenciales que probablemente motivan la compra en primer lugar. El hecho de que ciertos productos aparezcan recurrentemente como “primeros en el carrito” sugiere que son percibidos como indispensables en las compras semanales.
